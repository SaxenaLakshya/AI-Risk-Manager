```python
df["Product_Category"].value_counts()
```

Let's perform *Label Encoding* on column `Product_Category`.<br>
Fine, Label Encoding is not optimal for this task. It prioritize the ones with the higher numerical value, we shall go for the *One-hot Encoding*.

```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder()
```

```python
encoded = encoder.fit_transform(df[["Product_Category"]]).toarray()
encoded
```

```python
encoder_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out())
encoder_df
```

```python
df = pd.concat([df, encoder_df], axis=1)
```

```python
df.head()
```

```python
df.drop(["Product_Category"], axis=1, inplace=True)
df.head(5)
```

```python
df.info()
```

Honestly, when it comes to the Machine Learning Models. We must not be Gender Biased, Age can be factor affecting the return of the products.

```python
df.drop(["User_Gender"], axis=1, inplace=True)
df.info()
```

```python
df["Return_Status"].value_counts()
```

```python
df['Return_Status'] = df['Return_Status'].map({"Returned": 1, "Not Returned": 0}).astype(int)
df['Return_Status'].head(5)
```

```python
df['Return_Status'].isnull().sum().item()
```

```python
df.info()
```

```python
df['Return_Reason'].value_counts()
```

```python
df['Return_Reason'].isnull().sum()
```

Okay, so far I have observed is that we are predicting that will the product be returned by the customer based on the current information we have. Meaning, that the `Reason of Return` is something which is provided only if the product is returned.<br><br>
A little bit confusing, I know. We can collect this data for taking a look at the problems of the customers. Yes, we will use this for the predicting if the product will be returned by the customer or not. Since, this attribute us only present in the instances where the product is returned for sure, so it might make our model biased for this particular attribute, making it completely dependent on this faeture. So, we will remove this and use other features.

```python
df.drop(["Return_Reason"], axis=1, inplace=True)
df.info()
```

```python
df['Payment_Method'].value_counts()
```

```python
encoder = OneHotEncoder()
encoded = encoder.fit_transform(df[['Payment_Method']]).toarray()
encoded
```

```python
encoder_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out())
encoder_df
```

```python
df = pd.concat([df, encoder_df], axis=1)
df.head(5)
```

Another feature in the spotlight! `Days_to_Return`! This can also make our model biased that if this particular is present, then the product will be returned! Removing this can be much better for protecting the model from getting biased towards this feature.

```python
df.drop(["Days_to_Return", "Payment_Method"], axis=1, inplace=True)
df.info()
```

```python
df['Shipping_Method'].value_counts()
```

```python
encoder = OneHotEncoder()
encoded = encoder.fit_transform(df[['Shipping_Method']]).toarray()
encoded
```

```python
encoder_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out())
df = pd.concat([df, encoder_df], axis=1)
df.head(5)
```

```python
df.drop(["Shipping_Method"], axis=1, inplace=True)
df.head(5)
```

```python
df.info()
```
